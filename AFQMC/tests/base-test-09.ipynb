```python
L = 2
Nup = 1
Ndown = 1

t = 1.0
U = 4.0

dtau = 0.05
n_steps = 20000
n_walkers = 10
```

```python
lattice = Chain(n_sites=L, pbc=False)
hamiltonian = HubbardSystem(lattice=lattice, t=t, U=U)
propagator = HubbardPropagator(hamiltonian=hamiltonian, dtau=dtau)

trial_state = SlaterDeterminantTwoSpinState(hamiltonian=hamiltonian, n_electrons_up=Nup, n_electrons_down=Ndown)
trial_state.initialize("non-interacting")

walker = Walker(trial_state)
```

```python
x_arr = np.array([[1, 1],
              [1, -1],
              [-1, 1],
              [-1, -1]])


for i in range(n_steps):
    for x in x_arr:
        propagator.propagate(walker, x=x)
        walker.orthogonalize()


hamiltonian.calculate_variational_energy(walker)
```
